```python
import pickle
import numpy as np


with open('results.pkl', 'rb') as f:
    data = pickle.load(f)
```

```python
print(data.keys())
```

```python
print(np.shape(data['t']))
```

```python
#plot energies as a function of time

import matplotlib.pyplot as plt
plt.plot(data['t']*1j, data['energies_t'])
plt.xlabel('time')
plt.ylabel('energy')
plt.show()
```

```python
print(np.shape(data['psi_t']))
print(np.shape(data['psi_t'][:,0]))
```

```python
print(np.shape(data['x']))
print(data['x'][:,0,0])
```

```python
#plot the real part of the wavefunction as a function of space, for different times
plt.scatter(data['x'][:,0,0], data['psi_t'][:,0].real, label='t=0')
#plt.plot(data['x'][:,:,0], data['psi_t'][:,5].real, label='t=10')
#plt.scatter(data['x'][:,0,0], data['psi_t'][:,9].real, label='t=20')
plt.xlabel('space')
plt.ylabel('real part of wavefunction')
plt.legend()
plt.show()
```

```python
#plot the real part of the wavefunction as a function of space, for different times
#plt.scatter(data['x'][:,0,0], data['psi_t'][:,0].real, label='t=0')
#plt.plot(data['x'][:,:,0], data['psi_t'][:,5].real, label='t=10')
plt.scatter(data['x'][:,0,9], data['psi_t'][:,9].real, label='t=2')
plt.xlabel('space')
plt.ylabel('real part of wavefunction')
plt.legend()
plt.show()
```

```python
#plot a histogram of the walkers in space, pondered by the real part of the wavefunction
plt.hist(data['x'][:,0,9], weights=data['psi_t'][:,9].real, bins=500)
plt.xlabel('space')
plt.ylabel('weighted histogram of walkers')
plt.show()
```

```python
results = pickle.load(open("results.pkl", "rb"))
tau = -results['t'].imag
plt.plot(tau,
         results['energies_t'].real,
         label="Stochastic wavefunction result")
plt.plot(tau,
         3.0 * np.ones_like(tau),
         '--',
         label="Exact (2 noninteracting fermions in 2D)")
plt.xlabel(r'$\tau$')
plt.ylabel(r'$E_0$')
plt.legend()
plt.show()
```
