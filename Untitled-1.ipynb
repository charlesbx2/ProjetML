```python




def plot_laplacian_heatmap(results, time_step, coord_x=0, coord_y=1, n_bins=40):
    x_t      = results['x']        # (nsamples, dim, n_t)
    psi_t    = results['psi_t']    # (nsamples, n_t)
    d2psi_t  = results['d2psi_t']  # (nsamples, n_t)
    t        = results['t']
    tau      = -t.imag

    x_samples  = x_t[:, coord_x, time_step]
    y_samples  = x_t[:, coord_y, time_step]
    d2psi_vals = d2psi_t[:, time_step]
    psi_vals   = psi_t[:, time_step].real

    d2psi_clipped = np.clip(d2psi_vals,
                            np.percentile(d2psi_vals, 2),
                            np.percentile(d2psi_vals, 98))

    x_edges = np.linspace(x_samples.min(), x_samples.max(), n_bins + 1)
    y_edges = np.linspace(y_samples.min(), y_samples.max(), n_bins + 1)

    def bin_quantity(values):
        grid   = np.full((n_bins, n_bins), np.nan)
        counts = np.zeros((n_bins, n_bins))
        ix = np.digitize(x_samples, x_edges) - 1
        iy = np.digitize(y_samples, y_edges) - 1
        mask = (ix >= 0) & (ix < n_bins) & (iy >= 0) & (iy < n_bins)
        for k in np.where(mask)[0]:
            if np.isnan(grid[ix[k], iy[k]]):
                grid[ix[k], iy[k]] = 0.0
            grid[ix[k], iy[k]] += values[k]
            counts[ix[k], iy[k]] += 1
        nonzero = counts > 0
        grid[nonzero] /= counts[nonzero]
        return grid

    grid_d2psi = bin_quantity(d2psi_clipped)
    grid_psi   = bin_quantity(psi_vals)

    X, Y = np.meshgrid(
        0.5 * (x_edges[:-1] + x_edges[1:]),
        0.5 * (y_edges[:-1] + y_edges[1:]),
        indexing='ij'
    )

    coord_names = ['x₁', 'y₁', 'x₂', 'y₂']
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    fig.suptitle(f'Step i={time_step},  τ={tau[time_step]:.3f}', fontsize=13)

    im0 = axes[0].pcolormesh(X, Y, grid_psi, cmap='RdBu_r')
    axes[0].set_title('ψ')
    axes[0].set_xlabel(coord_names[coord_x])
    axes[0].set_ylabel(coord_names[coord_y])
    plt.colorbar(im0, ax=axes[0])

    im1 = axes[1].pcolormesh(X, Y, grid_d2psi, cmap='RdBu_r')
    axes[1].set_title('∇²ψ  (clampé p2–p98)')
    axes[1].set_xlabel(coord_names[coord_x])
    axes[1].set_ylabel(coord_names[coord_y])
    plt.colorbar(im1, ax=axes[1])

    plt.tight_layout()
    plt.show()

# --- Utilisation ---
plot_laplacian_heatmap(results,time_step=5, coord_x=0, coord_y=1)
```
