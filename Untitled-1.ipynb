```python




def plot_laplacian_heatmap(results, time_step, coord_x=0, coord_y=1, n_bins=40):
    x_t      = results['x']        # (nsamples, dim, n_t)
    psi_t    = results['psi_t']    # (nsamples, n_t)
    d2psi_t  = results['d2psi_t']  # (nsamples, n_t)
    t        = results['t']
    tau      = -t.imag

    x_samples  = x_t[:, coord_x, time_step]
    y_samples  = x_t[:, coord_y, time_step]
    d2psi_vals = d2psi_t[:, time_step]
    psi_vals   = psi_t[:, time_step].real

    d2psi_clipped = np.clip(d2psi_vals,
                            np.percentile(d2psi_vals, 2),
                            np.percentile(d2psi_vals, 98))

    x_edges = np.linspace(x_samples.min(), x_samples.max(), n_bins + 1)
    y_edges = np.linspace(y_samples.min(), y_samples.max(), n_bins + 1)

    def bin_quantity(values):
        grid   = np.full((n_bins, n_bins), np.nan)
        counts = np.zeros((n_bins, n_bins))
        ix = np.digitize(x_samples, x_edges) - 1
        iy = np.digitize(y_samples, y_edges) - 1
        mask = (ix >= 0) & (ix < n_bins) & (iy >= 0) & (iy < n_bins)
        for k in np.where(mask)[0]:
            if np.isnan(grid[ix[k], iy[k]]):
                grid[ix[k], iy[k]] = 0.0
            grid[ix[k], iy[k]] += values[k]
            counts[ix[k], iy[k]] += 1
        nonzero = counts > 0
        grid[nonzero] /= counts[nonzero]
        return grid

    grid_d2psi = bin_quantity(d2psi_clipped)
    grid_psi   = bin_quantity(psi_vals)

    X, Y = np.meshgrid(
        0.5 * (x_edges[:-1] + x_edges[1:]),
        0.5 * (y_edges[:-1] + y_edges[1:]),
        indexing='ij'
    )

    coord_names = ['x₁', 'y₁', 'x₂', 'y₂']
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    fig.suptitle(f'Step i={time_step},  τ={tau[time_step]:.3f}', fontsize=13)

    im0 = axes[0].pcolormesh(X, Y, grid_psi, cmap='RdBu_r')
    axes[0].set_title('ψ')
    axes[0].set_xlabel(coord_names[coord_x])
    axes[0].set_ylabel(coord_names[coord_y])
    plt.colorbar(im0, ax=axes[0])

    im1 = axes[1].pcolormesh(X, Y, grid_d2psi, cmap='RdBu_r')
    axes[1].set_title('∇²ψ  (clampé p2–p98)')
    axes[1].set_xlabel(coord_names[coord_x])
    axes[1].set_ylabel(coord_names[coord_y])
    plt.colorbar(im1, ax=axes[1])

    plt.tight_layout()
    plt.show()

# --- Utilisation ---
plot_laplacian_heatmap(results,time_step=5, coord_x=0, coord_y=1)
```

```python
# Diagnostic overfitting au dernier temps de convergence (200 epochs)

import glob
import os
import random
import itertools
import pickle
import numpy as np
import matplotlib.pyplot as plt
from sympy.combinatorics import Permutation

from params import (
    functionaltype, n_particles, dim_physical, perm_subset, bosonic, U,
    n_layers, layer_size, batch_size, reg, nsamples, GD
)

import stochastic_propagation_Nd as sp
if functionaltype == 1:
    import functional as nn_backend
else:
    import functional_neural_function_approximation_Nd as nn_backend

# 1) Récupérer les résultats déjà chargés, sinon charger le dernier fichier de résultats
if 'resultsGauss' in globals():
    results_ref = resultsGauss
elif 'results' in globals():
    results_ref = results
else:
    candidates = sorted(glob.glob('results*.pkl'), key=os.path.getmtime)
    if not candidates:
        raise FileNotFoundError("Aucun fichier results*.pkl trouvé dans le dossier courant.")
    with open(candidates[-1], 'rb') as f:
        results_ref = pickle.load(f)
    print(f"Fichier chargé: {candidates[-1]}")

# 2) Prendre le dernier pas de temps
last_t = results_ref['psi_t'].shape[1] - 15
samples_last = results_ref['x'][:, :, last_t]
psi_last = results_ref['psi_t'][:, last_t].real
psi_last = psi_last.reshape(-1, 1)

# Normalisation identique au code principal
scale = np.max(np.abs(psi_last))
if scale > 0:
    psi_last = psi_last / scale

# 3) Refaire l'augmentation par permutations comme dans la boucle de propagation
perm_all = list(itertools.permutations(range(n_particles)))
parity_all = [[Permutation(p).parity()] for p in perm_all]
subset = sorted(random.sample(range(len(parity_all)), min(perm_subset, len(parity_all))))
perm = [perm_all[i] for i in subset]
parity = [parity_all[i] for i in subset]

x_aug, y_aug = sp.permute(samples_last, psi_last, perm, parity, dim_physical, n_particles, bosonic)

# 4) Entraîner 200 epochs sur ce dernier état
n_epochs = 500
analysis_data = {}
fitfunc, d2fitfunc = nn_backend.neural_fit(
    x_aug,
    y_aug,
    n_samples=nsamples,
    perm_subset=perm_subset,
    perm=perm,
    parity=parity,
    analysis_data=analysis_data,
    iteration=9999,
    load_weights=0,
    bosonic=bosonic,
    U=U,
    n_particles=n_particles,
    dim_physical=dim_physical,
    n_layers=n_layers,
    layer_size=layer_size,
    epochs=n_epochs,
    batch_size=batch_size,
    reg=reg,
    GD=GD,
)

# 5) Extraire et tracer loss / val_loss
if functionaltype == 1:
    hist = analysis_data['history']
    train_loss = hist['loss']
    val_loss = hist['val_loss']
else:
    hist = analysis_data['history'].history
    train_loss = hist['loss']
    val_loss = hist['val_loss']

epochs_axis = np.arange(1, len(train_loss) + 1)
plt.figure(figsize=(8, 5))
plt.plot(epochs_axis, train_loss, label='train loss')
plt.plot(epochs_axis, val_loss, label='val loss')
plt.xlabel('epoch')
plt.ylabel('MSE loss')
plt.title(f'Overfitting check au dernier temps (t index = {last_t}, epochs = {n_epochs})')
plt.legend()
plt.grid(alpha=0.25)
plt.show()

print(f"Dernier index temps utilisé: {last_t}")
print(f"Train loss finale: {train_loss[-1]:.3e}")
print(f"Val loss finale:   {val_loss[-1]:.3e}")
```

```python
# Diagnostic overfitting au dernier temps de convergence
# Version demandée: results (pas resultsGauss), n_layers=2, layer_size=2, epochs=2000, GD=1

import itertools
import random
import numpy as np
import matplotlib.pyplot as plt
from sympy.combinatorics import Permutation

# 1) Forcer l'usage de results (sans priorité à resultsGauss)
if 'results' not in globals():
    raise NameError("La variable 'results' doit être définie dans le notebook.")
results_ref = results

# 2) Dernier pas de temps
last_t = results_ref['psi_t'].shape[1] - 1
samples_last = results_ref['x'][:, :, last_t]
psi_last = results_ref['psi_t'][:, last_t].real.reshape(-1, 1)

scale = np.max(np.abs(psi_last))
if scale > 0:
    psi_last = psi_last / scale

# 3) Augmentation par permutations
n_coords = samples_last.shape[1]
if n_coords % dim_physical != 0:
    raise ValueError(
        f"Incompatible: samples_last.shape[1]={n_coords} n'est pas divisible par dim_physical={dim_physical}."
    )

n_particles_data = n_coords // dim_physical
n_samples_data = samples_last.shape[0]
perm_subset_eff = min(perm_subset, np.math.factorial(n_particles_data))

perm_all = list(itertools.permutations(range(n_particles_data)))
parity_all = [[Permutation(p).parity()] for p in perm_all]
subset = sorted(random.sample(range(len(parity_all)), min(perm_subset_eff, len(parity_all))))
perm = [perm_all[i] for i in subset]
parity = [parity_all[i] for i in subset]

x_aug, y_aug = sp.permute(
    samples_last, psi_last, perm, parity, dim_physical, n_particles_data, bosonic
)

# 4) Entraînement avec hyperparamètres demandés
n_layers_test = 2
layer_size_test = 2*32
n_epochs = 100
GD_test = 1

analysis_data = {}
fitfunc, d2fitfunc = nn_backend.neural_fit(
    x_aug,
    y_aug,
    n_samples=n_samples_data,
    perm_subset=perm_subset_eff,
    perm=perm,
    parity=parity,
    analysis_data=analysis_data,
    iteration=9999,
    load_weights=0,
    bosonic=bosonic,
    U=U,
    n_particles=n_particles_data,
    dim_physical=dim_physical,
    n_layers=n_layers_test,
    layer_size=layer_size_test,
    epochs=n_epochs,
    batch_size=batch_size,
    reg=reg,
    GD=GD_test,
)

# 5) Tracé loss / val_loss
if functionaltype == 1:
    hist = analysis_data['history']
else:
    hist = analysis_data['history'].history

train_loss = hist['loss']
val_loss = hist['val_loss']

epochs_axis = np.arange(1, len(train_loss) + 1)
plt.figure(figsize=(8, 5))
plt.plot(epochs_axis, train_loss, label='train loss')
plt.plot(epochs_axis, val_loss, label='val loss')
plt.xlabel('epoch')
plt.ylabel('MSE loss')
plt.title(f'Overfitting check (results, t index={last_t}, epochs={n_epochs}, n_layers=2, layer_size=2, GD=1)')
plt.legend()
plt.grid(alpha=0.25)
plt.show()

print(f"Dernier index temps utilisé: {last_t}")
print(f"Train loss finale: {train_loss[-1]:.3e}")
print(f"Val loss finale:   {val_loss[-1]:.3e}")
```
